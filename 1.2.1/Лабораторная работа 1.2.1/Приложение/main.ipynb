```python
bullets = open('1.txt', 'r')
Am = [int(line) for line in bullets] # mg
bullets.close()

dm = 1 # mg
Em = [dm/m for m in Am] # доли
```

Отклонения:

```python
dev1 = open("3.txt", 'r')
Ad1 = [int(line) for line in dev1] # mkm
dev1.close()

dev2 = open("4.txt", 'r')
Ad2 = [int(line) for line in dev2] # 0.1 mm
dev2.close()

dd1 = 5 # mkm
Ed1 = [dd1/d for d in Ad1] # доли
dd2 = 5 # 0.1 mm
Ed2 = [dd2/d for d in Ad2] # доли
```

Параметры установок:

```python
g = 9.81

M = 295 # g
dM = 5 
eM = dM/M

L = 222 # cm
dL = 1
eL = dL/L


```

Первая установка:

```python
for i in range(5):
    m = Am[i]
    em = Em[i]
    x = Ad1[i]
    ex = Ed1[i]

    u = (M/m)*x*(g/L)**0.5
    u /= 10

    eu = eM**2 + em**2 + 0.25*eL**2 + ex**2
    eu = eu**0.5
    du = u*eu
    print(u, '+-', du, '***', eu)
```
